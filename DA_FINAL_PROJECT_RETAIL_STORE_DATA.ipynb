### Dataset Reloaded and missing values and datatypes finding

```python
import pandas as pd
import numpy as np

# Reload the dataset
df = pd.read_csv("https://raw.githubusercontent.com/Nithya2-S/Data_Analytics_Main_Project/refs/heads/main/retail_store_sales.csv")

# Quick check on missing values and data types
print(df.info())
print("\nMissing values per column:")
print(df.isnull().sum())
```

### Duplicates Checking
If Duplicates found then drop it or else keep the dataset as it is. As we already know from the initial EDA process ,there is no dupicates in this dataset.

```python
# Check for and drop duplicate rows, keeping the first occurrence
duplicate_count = df.duplicated().sum()
print(f"Number of duplicate rows found: {duplicate_count}")

if duplicate_count > 0:
    df.drop_duplicates(inplace=True)
    print(f"Removed {duplicate_count} duplicate rows. New shape: {df.shape}")
else:
    print("No duplicate rows found.")
```

### Datatype Changes

```python
# Datatype Changes

df["Transaction Date"] = pd.to_datetime(df["Transaction Date"])

df["Quantity"] = df["Quantity"].fillna(0).astype(int)

df['Discount Applied'] = df['Discount Applied'].astype(bool)

df.info()
```

### Handling Missing Values

Missing values were handled using relationship-based imputation. Missing Item values were inferred using the relationship between Category and Price Per Unit. Missing numerical values (Price Per Unit, Quantity, and Total Spent) were filled using the business rule Total Spent = Price Per Unit × Quantity to maintain data consistency.

#### 1.Fill Discount applied by Mode

```python
# Handling missing value of Discount Applied column
# Finding Mode

df["Discount Applied"].mode()
```

```python
# Fill the mode value for Discount Applied column

mode_value = df["Discount Applied"].mode()[0]
df["Discount Applied"] = df["Discount Applied"].fillna(value=mode_value)

print(df["Discount Applied"].isnull().sum())
```

#### 2.Fill Price Per Unit by Total spent/Quantity

```python
# Handling missing value of Price Per Unit column

df["Price Per Unit"] = df["Price Per Unit"].fillna(
    df["Total Spent"] / df["Quantity"]
)

print(df["Price Per Unit"].isnull().sum())
```

#### 3. Fill Item by Price per unit and category

```python
# Handling missing value of Item column

# Step 1: Create lookup table from existing non-missing Item rows
lookup = df.dropna(subset=["Item"])[
    ["Category", "Price Per Unit", "Item"]
].drop_duplicates()

# Step 2: Merge lookup back to original dataframe
df = df.merge(
    lookup,
    on=["Category", "Price Per Unit"],
    how="left",
    suffixes=("", "_filled")
)

# Step 3: Fill missing Item values
df["Item"] = df["Item"].fillna(df["Item_filled"])

# Step 4: Remove temporary column
df.drop("Item_filled", axis=1, inplace=True)

# Step 5: Check result
print(df["Item"].isnull().sum())
```

#### 4. Fill Quantity by mode within the Item

```python
# Handling missing value of Quantity column

df["Quantity"] = df.groupby("Item")["Quantity"].transform(
    lambda x: x.fillna(x.median())
)

print(df["Quantity"].isnull().sum())
```

#### 5. Fill Total Spent by Price Per Unit * Quantity

```python
# Handling missing value of Total Spent column

df["Total Spent"] = df["Total Spent"].fillna(
    df["Price Per Unit"] * df["Quantity"]
)

print(df["Item"].isnull().sum())
```

### Final Check for Info & missing values

```python
# Final Check Info & missing values:
df.info()
df.isnull().sum()
```

###Skewness checking
Skewness analysis was performed on numerical features. Price Per Unit and Quantity showed near-normal distributions with skewness values close to zero. Total Spent showed moderate positive skewness (0.826), but it remained within an acceptable range; therefore, no transformation was applied.

```python
# Check skewness score

initial_skew = df['Total Spent'].skew()

# Apply log transformation if skewness is high (absolute value > 1)
if abs(initial_skew) > 1:
    df['total_spent_log'] = np.log1p(df['total_spent'])
    print(f"High skewness ({initial_skew:.2f}) detected. Log transformation applied.")
else:
    print(f"Skewness is acceptable ({initial_skew:.2f}). No transformation needed.")
```

### Outliers finding

```python
# Check Outliers for all numeric column

for col in ["Price Per Unit", "Quantity", "Total Spent"]:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5*IQR
    upper = Q3 + 1.5*IQR
    outliers = df[(df[col] < lower) | (df[col] > upper)]
    print(f"{col}: {len(outliers)} outliers")
```

```python
# checking outlier rows for Total Spent

Q1 = df["Total Spent"].quantile(0.25)
Q3 = df["Total Spent"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df[(df["Total Spent"] < lower) | (df["Total Spent"] > upper)]
```

### Feature Transformations:

Feature transformations were applied to derive meaningful insights from the dataset. Date-based features such as Day of Week, Transaction Month, and Year were extracted from the transaction date. Additional derived fields including Customer Lifetime Value (CLV) and Purchase Frequency were created to analyze customer behavior and purchasing patterns.

```python
# Extract time features from the transaction date

df["Year"] = df["Transaction Date"].dt.year
df["Month"] = df["Transaction Date"].dt.month
df["Day_of_week"] = df["Transaction Date"].dt.day_name()

df['Is_weekend'] = df['Transaction Date'].dt.dayofweek.isin([5, 6]).astype(int)

df.head()
```

#### Check CLV & Customer Purchase Frequency

CLV(Customer Lifetime Value) was calculated by summing the total spending of each customer across all transactions.Purchase Frequency represents the total number of transactions made by each customer This metric helps identify high-value customers and analyze long-term customer contribution to revenue.

```python
# Customer Lifetime Value (CLV)
df["CLV"] = df.groupby("Customer ID")["Total Spent"].transform("sum")

# Purchase Frequency
df["Purchase_Frequency"] = df.groupby("Customer ID")["Transaction ID"].transform("count")

# Display results
print(df[["Customer ID", "Total Spent", "CLV", "Purchase_Frequency"]].head(10))
```

### Customer Summary

```python
# Create customer summary table
customer_summary = df.groupby("Customer ID").agg({
    "Total Spent": "sum",
    "Transaction ID": "count"
}).rename(columns={
    "Total Spent": "CLV",
    "Transaction ID": "Purchase_Frequency"
})

print(customer_summary)

# Customer with Highest CLV
highest_clv_customer = customer_summary.sort_values(
    by="CLV",
    ascending=False
).head(1)

# Customer with Highest Purchase Frequency
highest_pf_customer = customer_summary.sort_values(
    by="Purchase_Frequency",
    ascending=False
).head(1)

# Display results
print("Customer with Highest CLV:")
print(highest_clv_customer)

print("\nCustomer with Highest Purchase Frequency:")
print(highest_pf_customer)
```
